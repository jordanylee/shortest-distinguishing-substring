# SDS versus similarity
Four SVG exports total: agreement and Jaccard by alphabet (four panels each), and by state count (three panels each). Uses the current saved comparisons, with a common horizon read from their manifests.

Similarity is on x; mean finite SDS is on y. Equivalent pairs are excluded, zero lengths retained. These comparisons share DFAs and are descriptive. Logic is in `sds_similarity.py`.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / "data").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "dataviz"))
from sds_similarity import run, SOURCE_ID, RESULTS_ID

SOURCE_RUN = SOURCE_ID
COMPARISON_RUN = RESULTS_ID
OUTPUT = ROOT / "dataviz" / "figures" / "sds_vs_similarity"
BIN_COUNT = 20
```

```python
output = run(SOURCE_RUN, COMPARISON_RUN, OUTPUT, bin_count=BIN_COUNT)
```

```python
from IPython.display import SVG, display
FAMILIES = ("agreement", "jaccard")
for family in FAMILIES:
    for grouping in ("alphabets", "states"):
        display(SVG(filename=str(output / f"sds_vs_{family}_by_{grouping}.svg")))
```

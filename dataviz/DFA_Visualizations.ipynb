# DFA data visualization
Read-only analysis of saved comparisons. Edit the parameters and run all cells. The plotting/loading logic is in `analyze.py`. Outputs include a report, PNG/SVG figures, provenance, and CSV tables.

Pairs share DFAs; these plots are descriptive. Equivalent pairs have no finite SDS. Accepting-state fractions include unreachable states.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / "code").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "dataviz"))
from analyze import run, SOURCE, RESULT4, RESULT3

SOURCE_ID = SOURCE
H4_RESULTS_ID = RESULT4
H3_RESULTS_ID = RESULT3
OUTPUT = ROOT / "dataviz" / "outputs" / H4_RESULTS_ID
SEED = 42  # Recorded; full-data plots do not subsample.
```

```python
output = run(OUTPUT, SOURCE_ID, H4_RESULTS_ID, H3_RESULTS_ID, seed=SEED)
```

```python
import pandas as pd
from IPython.display import display, Markdown
summary = pd.read_csv(output / "tables" / "batch_summary.csv")
display(summary)
# Relative image links resolve in the saved report; display images explicitly below.
report = (output / "REPORT.md").read_text(encoding="utf-8")
display(Markdown(report.split("## Figure gallery")[0]))
```

```python
from IPython.display import Image
# Set a filename fragment to inspect selected figures, or leave empty for all.
FIGURE_FILTER = ""
for path in sorted(output.glob("*.png")):
    if FIGURE_FILTER in path.stem:
        display(Markdown(f"### {path.stem.replace('_', ' ')}"))
        display(Image(filename=str(path)))
```

```python
# Concrete rare witnesses, recurring DFA IDs, and bounded-identical pairs.
for name in ["tail_counts_and_dfas", "tail_dfa_involvement", "inequivalent_but_perfect_h4", "high_agreement_low_jaccard_summary"]:
    display(Markdown(f"### {name.replace('_', ' ')}"))
    display(pd.read_csv(output / "tables" / f"{name}.csv").head(25))
```
